````python
# ============================================================
# STEP 4 — DATASET STATISTICS
# ============================================================

import os
from PIL import Image
from collections import Counter

# Dataset path
DATASET_DIR = "/content/Agricultural_Land_Classification/data/extracted/images_dataSAT"

# Class folders
classes = {
    "class_0_non_agri": "Non-Agricultural",
    "class_1_agri": "Agricultural"
}

print("=" * 70)
print("🌾 SATELLITE AGRICULTURAL LAND DATASET")
print("=" * 70)

total_images = 0
image_sizes = Counter()
valid_images = 0
corrupted_images = []

# ------------------------------------------------------------
# Analyze each class
# ------------------------------------------------------------

for folder, class_name in classes.items():

    class_path = os.path.join(DATASET_DIR, folder)

    files = [
        f for f in os.listdir(class_path)
        if f.lower().endswith((".jpg", ".jpeg", ".png"))
    ]

    print(f"\n {class_name}")
    print(f"   Folder: {folder}")
    print(f"   Images: {len(files):,}")

    total_images += len(files)

    # Check image validity and dimensions
    for filename in files:

        filepath = os.path.join(class_path, filename)

        try:
            with Image.open(filepath) as img:
                img.verify()

            # Re-open after verify
            with Image.open(filepath) as img:
                image_sizes[img.size] += 1

            valid_images += 1

        except Exception:
            corrupted_images.append(filepath)

# ------------------------------------------------------------
# Results
# ------------------------------------------------------------

print("\n" + "=" * 70)
print(" DATASET SUMMARY")
print("=" * 70)

print(f"Total images       : {total_images:,}")
print(f"Valid images       : {valid_images:,}")
print(f"Corrupted images   : {len(corrupted_images):,}")

print("\n IMAGE DIMENSIONS")
print("-" * 70)

for size, count in image_sizes.most_common():
    print(f"{size[0]} × {size[1]}  →  {count:,} images")

print("\n⚖️ CLASS DISTRIBUTION")
print("-" * 70)

for folder, class_name in classes.items():

    class_path = os.path.join(DATASET_DIR, folder)

    count = len([
        f for f in os.listdir(class_path)
        if f.lower().endswith((".jpg", ".jpeg", ".png"))
    ])

    percentage = (count / total_images) * 100

    print(f"{class_name:20s}: {count:5,} images ({percentage:6.2f}%)")

print("\n" + "=" * 70)

if len(corrupted_images) == 0:
    print(" No corrupted images detected!")
else:
    print(f" {len(corrupted_images)} corrupted images detected.")

print("=" * 70)
````

````python
# ============================================================
# STEP 6 — ROBUST IMAGE DISCOVERY + DIMENSION CHECK
# ============================================================

import os
from PIL import Image
from collections import Counter

ROOT_DIR = "/content/Agricultural_Land_Classification/data/extracted"

IMAGE_EXTENSIONS = (".jpg", ".jpeg", ".png", ".tif", ".tiff")

image_records = []
dimension_counts = Counter()
corrupted_images = []

# ------------------------------------------------------------
# Find every image recursively
# ------------------------------------------------------------

for root, dirs, files in os.walk(ROOT_DIR):

    for filename in files:

        if filename.lower().endswith(IMAGE_EXTENSIONS):

            filepath = os.path.join(root, filename)

            try:
                with Image.open(filepath) as img:
                    img.load()

                    width, height = img.size
                    channels = len(img.getbands())

                    dimension_counts[(width, height, channels)] += 1

                    image_records.append({
                        "path": filepath,
                        "filename": filename,
                        "width": width,
                        "height": height,
                        "channels": channels
                    })

            except Exception as e:
                corrupted_images.append((filepath, str(e)))

# ------------------------------------------------------------
# Display results
# ------------------------------------------------------------

print("=" * 70)
print("IMAGE DATASET AUDIT")
print("=" * 70)

print(f"\n Total images found      : {len(image_records):,}")
print(f" Corrupted/unreadable    : {len(corrupted_images):,}")

print("\n Image dimensions:")
print("-" * 70)

for (width, height, channels), count in dimension_counts.most_common():

    print(
        f"{width} × {height} × {channels} "
        f"→ {count:,} images"
    )

print("\n" + "=" * 70)

if corrupted_images:
    print(" First corrupted files:")
    for filepath, error in corrupted_images[:5]:
        print(filepath)
else:
    print(" All discovered images can be opened successfully.")

print("=" * 70)
````

````python
# ============================================================
# STEP 11 — TRAIN / VALIDATION / TEST SPLIT
# ============================================================

import os
import pandas as pd
from sklearn.model_selection import train_test_split

# ------------------------------------------------------------
# Configuration
# ------------------------------------------------------------

BASE_DIR = "/content/Agricultural_Land_Classification/data/extracted/images_dataSAT"

CLASS_MAPPING = {
    "class_0_non_agri": 0,
    "class_1_agri": 1
}

CLASS_NAMES = {
    0: "Non-Agricultural",
    1: "Agricultural"
}

IMAGE_EXTENSIONS = (
    ".jpg",
    ".jpeg",
    ".png",
    ".tif",
    ".tiff"
)

# ------------------------------------------------------------
# Collect all images
# ------------------------------------------------------------

records = []

for folder_name, label in CLASS_MAPPING.items():

    class_dir = os.path.join(BASE_DIR, folder_name)

    if not os.path.exists(class_dir):
        print(f" Folder not found: {class_dir}")
        continue

    for root, dirs, files in os.walk(class_dir):

        for filename in files:

            if filename.lower().endswith(IMAGE_EXTENSIONS):

                filepath = os.path.join(root, filename)

                records.append({
                    "filepath": filepath,
                    "label": label,
                    "class_name": CLASS_NAMES[label]
                })

# ------------------------------------------------------------
# Create DataFrame
# ------------------------------------------------------------

df = pd.DataFrame(records)

print("=" * 75)
print("DATASET SUMMARY")
print("=" * 75)

print(f"\nTotal images: {len(df):,}")

print("\nClass distribution:")
print(df["class_name"].value_counts())

# ------------------------------------------------------------
# First split:
# 70% train
# 30% temporary
# ------------------------------------------------------------

train_df, temp_df = train_test_split(
    df,
    test_size=0.30,
    random_state=42,
    stratify=df["label"]
)

# ------------------------------------------------------------
# Second split:
# 15% validation
# 15% test
# ------------------------------------------------------------

val_df, test_df = train_test_split(
    temp_df,
    test_size=0.50,
    random_state=42,
    stratify=temp_df["label"]
)

# ------------------------------------------------------------
# Reset indices
# ------------------------------------------------------------

train_df = train_df.reset_index(drop=True)
val_df = val_df.reset_index(drop=True)
test_df = test_df.reset_index(drop=True)

# ------------------------------------------------------------
# Add split information
# ------------------------------------------------------------

train_df["split"] = "train"
val_df["split"] = "validation"
test_df["split"] = "test"

# Combined metadata
dataset_df = pd.concat(
    [train_df, val_df, test_df],
    ignore_index=True
)

# ------------------------------------------------------------
# Save metadata
# ------------------------------------------------------------

RESULTS_DIR = "/content/Agricultural_Land_Classification/results"

os.makedirs(RESULTS_DIR, exist_ok=True)

train_df.to_csv(
    os.path.join(RESULTS_DIR, "train_metadata.csv"),
    index=False
)

val_df.to_csv(
    os.path.join(RESULTS_DIR, "validation_metadata.csv"),
    index=False
)

test_df.to_csv(
    os.path.join(RESULTS_DIR, "test_metadata.csv"),
    index=False
)

# ------------------------------------------------------------
# Display split statistics
# ------------------------------------------------------------

print("\n" + "=" * 75)
print("DATA SPLIT")
print("=" * 75)

print(f"\nTraining set   : {len(train_df):,} images")
print(f"Validation set : {len(val_df):,} images")
print(f"Test set       : {len(test_df):,} images")

print(f"\nTotal          : {len(dataset_df):,} images")

print("\n" + "=" * 75)
print("CLASS DISTRIBUTION BY SPLIT")
print("=" * 75)

for split_name, split_df in [
    ("TRAIN", train_df),
    ("VALIDATION", val_df),
    ("TEST", test_df)
]:

    print(f"\n{split_name}")
    print("-" * 40)

    counts = split_df["class_name"].value_counts()

    for class_name, count in counts.items():

        percentage = (count / len(split_df)) * 100

        print(
            f"{class_name:20s}: "
            f"{count:5,} "
            f"({percentage:6.2f}%)"
        )

print("\n" + "=" * 75)
print(" TRAIN / VALIDATION / TEST SPLIT COMPLETE")
print("=" * 75)
````

````python
# ============================================================
# STEP 13 — KERAS DATA PIPELINE
# ============================================================

import os
import numpy as np
import tensorflow as tf
import matplotlib.pyplot as plt

print("TensorFlow version:", tf.__version__)

# ------------------------------------------------------------
# Configuration
# ------------------------------------------------------------

IMG_SIZE = (64, 64)
BATCH_SIZE = 32
AUTOTUNE = tf.data.AUTOTUNE

# ------------------------------------------------------------
# Make sure our DataFrames exist
# ------------------------------------------------------------

print("\nDataset sizes:")
print("Training   :", len(train_df))
print("Validation :", len(val_df))
print("Test       :", len(test_df))

# ------------------------------------------------------------
# Image loading function
# ------------------------------------------------------------

def load_image(filepath, label):

    # Read image file
    image = tf.io.read_file(filepath)

    # Decode JPEG
    image = tf.image.decode_jpeg(
        image,
        channels=3
    )

    # Resize to expected CNN input size
    image = tf.image.resize(
        image,
        IMG_SIZE
    )

    # Convert pixel values:
    # 0-255 → 0-1
    image = tf.cast(
        image,
        tf.float32
    ) / 255.0

    label = tf.cast(
        label,
        tf.float32
    )

    return image, label


# ------------------------------------------------------------
# Create tf.data.Dataset
# ------------------------------------------------------------

def create_dataset(dataframe, shuffle=False):

    paths = dataframe["filepath"].values
    labels = dataframe["label"].values.astype(np.float32)

    dataset = tf.data.Dataset.from_tensor_slices(
        (paths, labels)
    )

    if shuffle:

        dataset = dataset.shuffle(
            buffer_size=len(dataframe),
            seed=42,
            reshuffle_each_iteration=True
        )

    dataset = dataset.map(
        load_image,
        num_parallel_calls=AUTOTUNE
    )

    dataset = dataset.batch(
        BATCH_SIZE
    )

    dataset = dataset.prefetch(
        AUTOTUNE
    )

    return dataset


# ------------------------------------------------------------
# Create train / validation / test datasets
# ------------------------------------------------------------

train_ds = create_dataset(
    train_df,
    shuffle=True
)

val_ds = create_dataset(
    val_df,
    shuffle=False
)

test_ds = create_dataset(
    test_df,
    shuffle=False
)

print("\n Keras datasets created!")

print("\nDataset objects:")
print("Train      :", train_ds)
print("Validation :", val_ds)
print("Test       :", test_ds)
````

````python
# ============================================================
# STEP 14 — KERAS DATA AUGMENTATION
# ============================================================

import tensorflow as tf

data_augmentation = tf.keras.Sequential([

    tf.keras.layers.RandomFlip(
        mode="horizontal"
    ),

    tf.keras.layers.RandomRotation(
        factor=0.08
    ),

    tf.keras.layers.RandomZoom(
        height_factor=0.10,
        width_factor=0.10
    ),

], name="data_augmentation")

print(" Augmentation pipeline created!")

data_augmentation.summary()
````

````python
# ============================================================
# STEP 14B — VISUALIZE AUGMENTATION
# ============================================================

# Take one batch
images, labels = next(iter(train_ds))

# Select one image
original_image = images[0]

# Create augmented versions
augmented_images = []

for _ in range(5):

    augmented = data_augmentation(
        tf.expand_dims(original_image, axis=0),
        training=True
    )

    augmented_images.append(
        augmented[0].numpy()
    )

# ------------------------------------------------------------
# Plot
# ------------------------------------------------------------

fig, axes = plt.subplots(
    2,
    3,
    figsize=(10, 7)
)

# Original
axes[0, 0].imshow(original_image.numpy())
axes[0, 0].set_title("Original")
axes[0, 0].axis("off")

# Augmented versions
for i, augmented_image in enumerate(augmented_images):

    row = (i + 1) // 3
    col = (i + 1) % 3

    axes[row, col].imshow(
        np.clip(augmented_image, 0, 1)
    )

    axes[row, col].set_title(
        f"Augmented {i + 1}"
    )

    axes[row, col].axis("off")

plt.suptitle(
    "On-the-Fly Data Augmentation",
    fontsize=15,
    fontweight="bold"
)

plt.tight_layout()
plt.show()
````

````python
# ============================================================
# STEP 15 — BUILD KERAS CNN
# ============================================================

import tensorflow as tf

# ------------------------------------------------------------
# Configuration
# ------------------------------------------------------------

IMG_HEIGHT = 64
IMG_WIDTH = 64
CHANNELS = 3

# ------------------------------------------------------------
# Build CNN
# ------------------------------------------------------------

keras_cnn = tf.keras.Sequential([

    # Input
    tf.keras.layers.Input(
        shape=(IMG_HEIGHT, IMG_WIDTH, CHANNELS)
    ),

    # Data augmentation
    data_augmentation,

    # --------------------------------------------------------
    # Convolution Block 1
    # --------------------------------------------------------

    tf.keras.layers.Conv2D(
        filters=32,
        kernel_size=(3, 3),
        padding="same",
        activation="relu"
    ),

    tf.keras.layers.BatchNormalization(),

    tf.keras.layers.MaxPooling2D(
        pool_size=(2, 2)
    ),

    # --------------------------------------------------------
    # Convolution Block 2
    # --------------------------------------------------------

    tf.keras.layers.Conv2D(
        filters=64,
        kernel_size=(3, 3),
        padding="same",
        activation="relu"
    ),

    tf.keras.layers.BatchNormalization(),

    tf.keras.layers.MaxPooling2D(
        pool_size=(2, 2)
    ),

    # --------------------------------------------------------
    # Convolution Block 3
    # --------------------------------------------------------

    tf.keras.layers.Conv2D(
        filters=128,
        kernel_size=(3, 3),
        padding="same",
        activation="relu"
    ),

    tf.keras.layers.BatchNormalization(),

    tf.keras.layers.MaxPooling2D(
        pool_size=(2, 2)
    ),

    # --------------------------------------------------------
    # Regularization
    # --------------------------------------------------------

    tf.keras.layers.Dropout(0.30),

    # --------------------------------------------------------
    # Classification head
    # --------------------------------------------------------

    tf.keras.layers.GlobalAveragePooling2D(),

    tf.keras.layers.Dense(
        128,
        activation="relu"
    ),

    tf.keras.layers.Dropout(0.40),

    # Binary classification
    tf.keras.layers.Dense(
        1,
        activation="sigmoid"
    )
], name="Keras_CNN")

# ------------------------------------------------------------
# Compile
# ------------------------------------------------------------

keras_cnn.compile(
    optimizer=tf.keras.optimizers.Adam(
        learning_rate=1e-3
    ),

    loss="binary_crossentropy",

    metrics=[
        "accuracy",
        tf.keras.metrics.Precision(name="precision"),
        tf.keras.metrics.Recall(name="recall"),
        tf.keras.metrics.AUC(name="auc")
    ]
)

# ------------------------------------------------------------
# Model summary
# ------------------------------------------------------------

keras_cnn.summary()
````

````python
# ============================================================
# STEP 20A — PYTORCH SETUP AND TRANSFORMS
# ============================================================

import os
import numpy as np
import torch
import torchvision
import matplotlib.pyplot as plt

from PIL import Image
from torch.utils.data import Dataset, DataLoader
from torchvision import transforms

# ------------------------------------------------------------
# Device
# ------------------------------------------------------------

device = torch.device(
    "cuda" if torch.cuda.is_available() else "cpu"
)

print("=" * 70)
print("PYTORCH SETUP")
print("=" * 70)

print("PyTorch version :", torch.__version__)
print("Torchvision     :", torchvision.__version__)
print("Device          :", device)

if torch.cuda.is_available():
    print("GPU             :", torch.cuda.get_device_name(0))

# ------------------------------------------------------------
# Image configuration
# ------------------------------------------------------------

IMG_SIZE = 64
BATCH_SIZE = 32

# ------------------------------------------------------------
# Training transformations
# ------------------------------------------------------------

train_transform = transforms.Compose([

    transforms.Resize(
        (IMG_SIZE, IMG_SIZE)
    ),

    transforms.RandomHorizontalFlip(
        p=0.5
    ),

    transforms.RandomRotation(
        degrees=10
    ),

    transforms.RandomResizedCrop(
        size=IMG_SIZE,
        scale=(0.90, 1.0)
    ),

    transforms.ToTensor(),

])

# ------------------------------------------------------------
# Validation/Test transformations
# ------------------------------------------------------------

eval_transform = transforms.Compose([

    transforms.Resize(
        (IMG_SIZE, IMG_SIZE)
    ),

    transforms.ToTensor(),

])

print("\n PyTorch transforms created!")

print("\nTraining transform:")
print(train_transform)

print("\nEvaluation transform:")
print(eval_transform)
````

````python
# ============================================================
# STEP 20B — PYTORCH CUSTOM DATASET
# ============================================================

class AgriculturalLandDataset(Dataset):

    def __init__(self, dataframe, transform=None):

        self.dataframe = dataframe.reset_index(drop=True)
        self.transform = transform

    def __len__(self):

        return len(self.dataframe)

    def __getitem__(self, index):

        # Get image path
        image_path = self.dataframe.loc[
            index,
            "filepath"
        ]

        # Get label
        label = int(
            self.dataframe.loc[
                index,
                "label"
            ]
        )

        # Open image
        image = Image.open(
            image_path
        ).convert("RGB")

        # Apply transformations
        if self.transform is not None:

            image = self.transform(image)

        # PyTorch classification labels
        label = torch.tensor(
            label,
            dtype=torch.float32
        )

        return image, label


# ------------------------------------------------------------
# Create datasets
# ------------------------------------------------------------

train_dataset = AgriculturalLandDataset(
    train_df,
    transform=train_transform
)

val_dataset = AgriculturalLandDataset(
    val_df,
    transform=eval_transform
)

test_dataset = AgriculturalLandDataset(
    test_df,
    transform=eval_transform
)

print("=" * 70)
print("PYTORCH DATASETS")
print("=" * 70)

print("\nTraining images   :", len(train_dataset))
print("Validation images :", len(val_dataset))
print("Test images       :", len(test_dataset))

print("\n PyTorch Dataset objects created!")
````

````python
# ============================================================
# STEP 20C — PYTORCH DATALOADERS
# ============================================================

train_loader = DataLoader(
    train_dataset,
    batch_size=BATCH_SIZE,
    shuffle=True,
    num_workers=2,
    pin_memory=True
)

val_loader = DataLoader(
    val_dataset,
    batch_size=BATCH_SIZE,
    shuffle=False,
    num_workers=2,
    pin_memory=True
)

test_loader = DataLoader(
    test_dataset,
    batch_size=BATCH_SIZE,
    shuffle=False,
    num_workers=2,
    pin_memory=True
)

print("=" * 70)
print("PYTORCH DATALOADERS")
print("=" * 70)

print("\nTraining batches   :", len(train_loader))
print("Validation batches :", len(val_loader))
print("Test batches       :", len(test_loader))

print("\nBatch size         :", BATCH_SIZE)

print("\n DataLoaders ready!")
````

````python
# ============================================================
# STEP 21 — PYTORCH CNN MODEL
# ============================================================

import torch
import torch.nn as nn

# ============================================================
# CNN ARCHITECTURE
# ============================================================

class AgriculturalCNN(nn.Module):

    def __init__(self):

        super().__init__()

        # ----------------------------------------------------
        # Convolutional feature extractor
        # ----------------------------------------------------

        self.features = nn.Sequential(

            # Block 1
            nn.Conv2d(
                in_channels=3,
                out_channels=32,
                kernel_size=3,
                padding=1
            ),

            nn.BatchNorm2d(32),

            nn.ReLU(),

            nn.MaxPool2d(2),

            # 64x64 -> 32x32


            # Block 2
            nn.Conv2d(
                in_channels=32,
                out_channels=64,
                kernel_size=3,
                padding=1
            ),

            nn.BatchNorm2d(64),

            nn.ReLU(),

            nn.MaxPool2d(2),

            # 32x32 -> 16x16


            # Block 3
            nn.Conv2d(
                in_channels=64,
                out_channels=128,
                kernel_size=3,
                padding=1
            ),

            nn.BatchNorm2d(128),

            nn.ReLU(),

            nn.MaxPool2d(2)

            # 16x16 -> 8x8
        )

        # ----------------------------------------------------
        # Global Average Pooling
        # ----------------------------------------------------

        self.global_pool = nn.AdaptiveAvgPool2d(
            (1, 1)
        )

        # ----------------------------------------------------
        # Classification head
        # ----------------------------------------------------

        self.classifier = nn.Sequential(

            nn.Flatten(),

            nn.Linear(
                128,
                128
            ),

            nn.ReLU(),

            nn.Dropout(0.40),

            # One output for binary classification
            nn.Linear(
                128,
                1
            )
        )

    def forward(self, x):

        x = self.features(x)

        x = self.global_pool(x)

        x = self.classifier(x)

        return x


# ============================================================
# CREATE MODEL
# ============================================================

pytorch_cnn = AgriculturalCNN()

# Move model to GPU
pytorch_cnn = pytorch_cnn.to(device)

# ============================================================
# LOSS + OPTIMIZER
# ============================================================

criterion = nn.BCEWithLogitsLoss()

optimizer = torch.optim.Adam(
    pytorch_cnn.parameters(),
    lr=1e-3
)

# ============================================================
# MODEL INFORMATION
# ============================================================

print("=" * 70)
print("PYTORCH CNN")
print("=" * 70)

print(pytorch_cnn)

# Count parameters
total_parameters = sum(
    p.numel()
    for p in pytorch_cnn.parameters()
)

trainable_parameters = sum(
    p.numel()
    for p in pytorch_cnn.parameters()
    if p.requires_grad
)

print("\n" + "=" * 70)

print(f"Total parameters      : {total_parameters:,}")
print(f"Trainable parameters  : {trainable_parameters:,}")

print("\nDevice:")
print(device)

print("\n PyTorch CNN created successfully!")
````

````python
# ============================================================
# STEP 27A — VISION TRANSFORMER PATCH EXTRACTION
# ============================================================

import tensorflow as tf
import matplotlib.pyplot as plt
import numpy as np

# ------------------------------------------------------------
# Configuration
# ------------------------------------------------------------

IMAGE_SIZE = 64
PATCH_SIZE = 8
NUM_PATCHES = (IMAGE_SIZE // PATCH_SIZE) ** 2

print("=" * 70)
print("VISION TRANSFORMER PATCH CONFIGURATION")
print("=" * 70)

print(f"\nImage size     : {IMAGE_SIZE} × {IMAGE_SIZE}")
print(f"Patch size     : {PATCH_SIZE} × {PATCH_SIZE}")
print(f"Number patches : {NUM_PATCHES}")

# ------------------------------------------------------------
# Patch extraction function
# ------------------------------------------------------------

def extract_patches(images):

    patches = tf.image.extract_patches(
        images=images,
        sizes=[
            1,
            PATCH_SIZE,
            PATCH_SIZE,
            1
        ],
        strides=[
            1,
            PATCH_SIZE,
            PATCH_SIZE,
            1
        ],
        rates=[
            1,
            1,
            1,
            1
        ],
        padding="VALID"
    )

    # Each patch is flattened
    patch_dims = patches.shape[-1]

    patches = tf.reshape(
        patches,
        [
            tf.shape(images)[0],
            NUM_PATCHES,
            patch_dims
        ]
    )

    return patches


# ------------------------------------------------------------
# Test using one batch
# ------------------------------------------------------------

images, labels = next(iter(train_ds))

patches = extract_patches(images)

print("\nOriginal batch:")
print(images.shape)

print("\nExtracted patches:")
print(patches.shape)

print("\n" + "=" * 70)
print(" PATCH EXTRACTION WORKING")
print("=" * 70)
````

````python
# ============================================================
# STEP 28 — KERAS VISION TRANSFORMER
# ============================================================

import tensorflow as tf
from tensorflow import keras
from tensorflow.keras import layers

# ------------------------------------------------------------
# Configuration
# ------------------------------------------------------------

IMAGE_SIZE = 64
PATCH_SIZE = 8
NUM_PATCHES = 64

PROJECTION_DIM = 64
NUM_HEADS = 4

TRANSFORMER_LAYERS = 3

TRANSFORMER_UNITS = [
    PROJECTION_DIM * 2,
    PROJECTION_DIM
]

MLP_HEAD_UNITS = [
    128,
    64
]


# ============================================================
# Patch Encoder
# ============================================================

class PatchEncoder(layers.Layer):

    def __init__(
        self,
        num_patches,
        projection_dim
    ):

        super().__init__()

        self.num_patches = num_patches

        self.projection = layers.Dense(
            units=projection_dim
        )

        self.position_embedding = layers.Embedding(
            input_dim=num_patches,
            output_dim=projection_dim
        )


    def call(self, patch):

        positions = tf.range(
            start=0,
            limit=self.num_patches,
            delta=1
        )

        encoded = (
            self.projection(patch)
            +
            self.position_embedding(positions)
        )

        return encoded


# ============================================================
# Transformer Block
# ============================================================

def transformer_block(
    x,
    projection_dim,
    num_heads,
    transformer_units
):

    # --------------------------------------------------------
    # Layer normalization
    # --------------------------------------------------------

    x1 = layers.LayerNormalization(
        epsilon=1e-6
    )(x)

    # --------------------------------------------------------
    # Multi-head self attention
    # --------------------------------------------------------

    attention_output = layers.MultiHeadAttention(
        num_heads=num_heads,
        key_dim=projection_dim // num_heads,
        dropout=0.10
    )(
        x1,
        x1
    )

    # --------------------------------------------------------
    # Residual connection
    # --------------------------------------------------------

    x2 = layers.Add()([
        attention_output,
        x
    ])

    # --------------------------------------------------------
    # MLP
    # --------------------------------------------------------

    x3 = layers.LayerNormalization(
        epsilon=1e-6
    )(x2)

    for units in transformer_units:

        x3 = layers.Dense(
            units,
            activation=tf.nn.gelu
        )(x3)

        x3 = layers.Dropout(
            0.10
        )(x3)

    # --------------------------------------------------------
    # Second residual connection
    # --------------------------------------------------------

    output = layers.Add()([
        x3,
        x2
    ])

    return output


# ============================================================
# Build Vision Transformer
# ============================================================

inputs = layers.Input(
    shape=(IMAGE_SIZE, IMAGE_SIZE, 3)
)

# ------------------------------------------------------------
# Data augmentation
# ------------------------------------------------------------

x = data_augmentation(inputs)

# ------------------------------------------------------------
# Extract patches
# ------------------------------------------------------------

patches = layers.Lambda(
    extract_patches,
    name="patch_extraction"
)(x)

# ------------------------------------------------------------
# Patch embedding + positional embedding
# ------------------------------------------------------------

x = PatchEncoder(
    NUM_PATCHES,
    PROJECTION_DIM
)(patches)

# ------------------------------------------------------------
# Transformer encoder blocks
# ------------------------------------------------------------

for i in range(TRANSFORMER_LAYERS):

    x = transformer_block(
        x,
        PROJECTION_DIM,
        NUM_HEADS,
        TRANSFORMER_UNITS
    )

# ------------------------------------------------------------
# Classification head
# ------------------------------------------------------------

x = layers.LayerNormalization(
    epsilon=1e-6
)(x)

x = layers.GlobalAveragePooling1D()(x)

for units in MLP_HEAD_UNITS:

    x = layers.Dense(
        units,
        activation=tf.nn.gelu
    )(x)

    x = layers.Dropout(
        0.30
    )(x)

outputs = layers.Dense(
    1,
    activation="sigmoid"
)(x)


# ------------------------------------------------------------
# Create model
# ------------------------------------------------------------

keras_vit = keras.Model(
    inputs=inputs,
    outputs=outputs,
    name="Keras_Vision_Transformer"
)


# ============================================================
# Compile
# ============================================================

keras_vit.compile(

    optimizer=keras.optimizers.AdamW(
        learning_rate=3e-4,
        weight_decay=1e-4
    ),

    loss="binary_crossentropy",

    metrics=[
        "accuracy",

        keras.metrics.Precision(
            name="precision"
        ),

        keras.metrics.Recall(
            name="recall"
        ),

        keras.metrics.AUC(
            name="auc"
        )
    ]
)


# ============================================================
# Summary
# ============================================================

print("=" * 70)
print("KERAS VISION TRANSFORMER")
print("=" * 70)

keras_vit.summary()

print("\n" + "=" * 70)
print(" KERAS ViT CREATED SUCCESSFULLY")
print("=" * 70)
````

````python
# ============================================================
# STEP 33 — PYTORCH VISION TRANSFORMER
# ============================================================

import torch
import torch.nn as nn


# ============================================================
# CONFIGURATION
# ============================================================

IMAGE_SIZE = 64
PATCH_SIZE = 8

NUM_CHANNELS = 3

NUM_PATCHES = (
    IMAGE_SIZE // PATCH_SIZE
) ** 2

EMBED_DIM = 64

NUM_HEADS = 4

NUM_LAYERS = 3

MLP_DIM = 128

DROPOUT = 0.10


# ============================================================
# PATCH EMBEDDING
# ============================================================

class PatchEmbedding(nn.Module):

    def __init__(
        self,
        image_size,
        patch_size,
        in_channels,
        embed_dim
    ):

        super().__init__()

        self.image_size = image_size

        self.patch_size = patch_size

        self.num_patches = (
            image_size // patch_size
        ) ** 2

        # Conv2D performs:
        #
        # image → non-overlapping patches
        # patches → embeddings

        self.projection = nn.Conv2d(
            in_channels=in_channels,
            out_channels=embed_dim,
            kernel_size=patch_size,
            stride=patch_size
        )


    def forward(self, x):

        # ----------------------------------------------------
        # Input:
        #
        # [batch, 3, 64, 64]
        # ----------------------------------------------------

        x = self.projection(x)

        # ----------------------------------------------------
        # After projection:
        #
        # [batch, embed_dim, 8, 8]
        # ----------------------------------------------------

        # Flatten spatial dimensions

        x = x.flatten(
            2
        )

        # ----------------------------------------------------
        # [batch, embed_dim, 64]
        # ----------------------------------------------------

        # Move embedding dimension to the end

        x = x.transpose(
            1,
            2
        )

        # ----------------------------------------------------
        # [batch, 64, embed_dim]
        # ----------------------------------------------------

        return x


# ============================================================
# PYTORCH VISION TRANSFORMER
# ============================================================

class AgriculturalViT(nn.Module):

    def __init__(self):

        super().__init__()

        # ----------------------------------------------------
        # Patch embedding
        # ----------------------------------------------------

        self.patch_embedding = PatchEmbedding(
            image_size=IMAGE_SIZE,
            patch_size=PATCH_SIZE,
            in_channels=NUM_CHANNELS,
            embed_dim=EMBED_DIM
        )

        # ----------------------------------------------------
        # Learnable class token
        # ----------------------------------------------------

        self.cls_token = nn.Parameter(
            torch.zeros(
                1,
                1,
                EMBED_DIM
            )
        )

        # ----------------------------------------------------
        # Positional embedding
        #
        # +1 because of CLS token
        # ----------------------------------------------------

        self.position_embedding = nn.Parameter(
            torch.zeros(
                1,
                NUM_PATCHES + 1,
                EMBED_DIM
            )
        )

        # ----------------------------------------------------
        # Transformer encoder
        # ----------------------------------------------------

        encoder_layer = (
            nn.TransformerEncoderLayer(
                d_model=EMBED_DIM,
                nhead=NUM_HEADS,
                dim_feedforward=MLP_DIM,
                dropout=DROPOUT,
                activation="gelu",
                batch_first=True,
                norm_first=True
            )
        )

        self.transformer = (
            nn.TransformerEncoder(
                encoder_layer,
                num_layers=NUM_LAYERS
            )
        )

        # ----------------------------------------------------
        # Final normalization
        # ----------------------------------------------------

        self.norm = nn.LayerNorm(
            EMBED_DIM
        )

        # ----------------------------------------------------
        # Classification head
        # ----------------------------------------------------

        self.classifier = nn.Sequential(

            nn.Linear(
                EMBED_DIM,
                64
            ),

            nn.GELU(),

            nn.Dropout(
                0.30
            ),

            nn.Linear(
                64,
                1
            )
        )


    def forward(self, x):

        # ----------------------------------------------------
        # Patch embeddings
        # ----------------------------------------------------

        x = self.patch_embedding(x)

        # ----------------------------------------------------
        # Add CLS token
        # ----------------------------------------------------

        batch_size = x.size(0)

        cls_tokens = self.cls_token.expand(
            batch_size,
            -1,
            -1
        )

        x = torch.cat(
            [
                cls_tokens,
                x
            ],
            dim=1
        )

        # ----------------------------------------------------
        # Add positional information
        # ----------------------------------------------------

        x = (
            x
            +
            self.position_embedding
        )

        # ----------------------------------------------------
        # Transformer
        # ----------------------------------------------------

        x = self.transformer(
            x
        )

        # ----------------------------------------------------
        # Take CLS token
        # ----------------------------------------------------

        x = x[:, 0]

        # ----------------------------------------------------
        # Normalize
        # ----------------------------------------------------

        x = self.norm(
            x
        )

        # ----------------------------------------------------
        # Classification
        # ----------------------------------------------------

        x = self.classifier(
            x
        )

        return x


# ============================================================
# CREATE MODEL
# ============================================================

pytorch_vit = AgriculturalViT()

pytorch_vit = pytorch_vit.to(
    device
)


# ============================================================
# LOSS
# ============================================================

vit_criterion = nn.BCEWithLogitsLoss()


# ============================================================
# OPTIMIZER
# ============================================================

vit_optimizer = torch.optim.AdamW(
    pytorch_vit.parameters(),
    lr=3e-4,
    weight_decay=1e-4
)


# ============================================================
# MODEL INFORMATION
# ============================================================

total_parameters = sum(
    p.numel()
    for p in pytorch_vit.parameters()
)

trainable_parameters = sum(
    p.numel()
    for p in pytorch_vit.parameters()
    if p.requires_grad
)


print("=" * 80)
print("PYTORCH VISION TRANSFORMER")
print("=" * 80)

print(
    f"\nImage size       : {IMAGE_SIZE} × {IMAGE_SIZE}"
)

print(
    f"Patch size       : {PATCH_SIZE} × {PATCH_SIZE}"
)

print(
    f"Number of patches: {NUM_PATCHES}"
)

print(
    f"Embedding dim    : {EMBED_DIM}"
)

print(
    f"Attention heads  : {NUM_HEADS}"
)

print(
    f"Transformer layers: {NUM_LAYERS}"
)

print(
    f"\nTotal parameters : {total_parameters:,}"
)

print(
    f"Trainable params : {trainable_parameters:,}"
)

print(
    f"Device           : {device}"
)

print("\n" + "=" * 80)
print(pytorch_vit)
print("=" * 80)

print(
    "\n PyTorch Vision Transformer created successfully!"
)
````

````python
# ============================================================
# STEP 35 — TRAIN PYTORCH VISION TRANSFORMER
# ============================================================

import os
import time
import numpy as np
import torch

# ============================================================
# CONFIGURATION
# ============================================================

EPOCHS = 20

PATIENCE = 5

BASE_DIR = "/content/Agricultural_Land_Classification"

MODEL_DIR = os.path.join(
    BASE_DIR,
    "models"
)

RESULTS_DIR = os.path.join(
    BASE_DIR,
    "results"
)

METRICS_DIR = os.path.join(
    RESULTS_DIR,
    "metrics"
)

FIGURES_DIR = os.path.join(
    RESULTS_DIR,
    "figures"
)

os.makedirs(MODEL_DIR, exist_ok=True)
os.makedirs(METRICS_DIR, exist_ok=True)
os.makedirs(FIGURES_DIR, exist_ok=True)

BEST_VIT_PATH = os.path.join(
    MODEL_DIR,
    "pytorch_vit_best.pth"
)


# ============================================================
# HISTORY
# ============================================================

pytorch_vit_history = {
    "train_loss": [],
    "val_loss": [],
    "train_accuracy": [],
    "val_accuracy": []
}


# ============================================================
# BEST MODEL TRACKING
# ============================================================

best_val_loss = float("inf")

patience_counter = 0


# ============================================================
# LEARNING RATE SCHEDULER
# ============================================================

vit_scheduler = torch.optim.lr_scheduler.ReduceLROnPlateau(
    vit_optimizer,
    mode="min",
    factor=0.5,
    patience=2
)


# ============================================================
# TRAINING START
# ============================================================

print("=" * 80)
print("PYTORCH VISION TRANSFORMER TRAINING")
print("=" * 80)

print(f"\nDevice       : {device}")
print(f"Epochs       : {EPOCHS}")
print(f"Early stop   : {PATIENCE} epochs")
print(f"Model output : Binary classification")

print("\n" + "=" * 80)


# ============================================================
# EPOCH LOOP
# ============================================================

for epoch in range(EPOCHS):

    epoch_start = time.time()


    # ========================================================
    # TRAINING
    # ========================================================

    pytorch_vit.train()

    train_loss_total = 0.0

    train_correct = 0

    train_total = 0


    for images, labels in train_loader:

        # ----------------------------------------------------
        # Move data to GPU
        # ----------------------------------------------------

        images = images.to(
            device,
            non_blocking=True
        )

        labels = labels.to(
            device,
            non_blocking=True
        )

        # ----------------------------------------------------
        # Shape labels
        # [batch] → [batch, 1]
        # ----------------------------------------------------

        labels = labels.float().unsqueeze(1)


        # ----------------------------------------------------
        # Clear gradients
        # ----------------------------------------------------

        vit_optimizer.zero_grad()


        # ----------------------------------------------------
        # Forward pass
        # ----------------------------------------------------

        outputs = pytorch_vit(
            images
        )


        # ----------------------------------------------------
        # Calculate loss
        # ----------------------------------------------------

        loss = vit_criterion(
            outputs,
            labels
        )


        # ----------------------------------------------------
        # Backpropagation
        # ----------------------------------------------------

        loss.backward()


        # ----------------------------------------------------
        # Update weights
        # ----------------------------------------------------

        vit_optimizer.step()


        # ----------------------------------------------------
        # Statistics
        # ----------------------------------------------------

        batch_size = labels.size(0)

        train_loss_total += (
            loss.item() * batch_size
        )


        # Convert logits to probabilities

        probabilities = torch.sigmoid(
            outputs
        )


        # Convert probabilities to classes

        predictions = (
            probabilities >= 0.5
        ).float()


        train_correct += (
            predictions == labels
        ).sum().item()


        train_total += batch_size


    # ========================================================
    # TRAINING METRICS
    # ========================================================

    train_loss = (
        train_loss_total /
        train_total
    )

    train_accuracy = (
        train_correct /
        train_total
    )


    # ========================================================
    # VALIDATION
    # ========================================================

    pytorch_vit.eval()

    val_loss_total = 0.0

    val_correct = 0

    val_total = 0


    with torch.no_grad():

        for images, labels in val_loader:

            # ------------------------------------------------
            # Move data
            # ------------------------------------------------

            images = images.to(
                device,
                non_blocking=True
            )

            labels = labels.to(
                device,
                non_blocking=True
            )

            labels = labels.float().unsqueeze(1)


            # ------------------------------------------------
            # Forward pass
            # ------------------------------------------------

            outputs = pytorch_vit(
                images
            )


            # ------------------------------------------------
            # Validation loss
            # ------------------------------------------------

            loss = vit_criterion(
                outputs,
                labels
            )


            batch_size = labels.size(0)

            val_loss_total += (
                loss.item() * batch_size
            )


            # ------------------------------------------------
            # Predictions
            # ------------------------------------------------

            probabilities = torch.sigmoid(
                outputs
            )

            predictions = (
                probabilities >= 0.5
            ).float()


            val_correct += (
                predictions == labels
            ).sum().item()


            val_total += batch_size


    # ========================================================
    # VALIDATION METRICS
    # ========================================================

    val_loss = (
        val_loss_total /
        val_total
    )

    val_accuracy = (
        val_correct /
        val_total
    )


    # ========================================================
    # STORE HISTORY
    # ========================================================

    pytorch_vit_history[
        "train_loss"
    ].append(
        train_loss
    )

    pytorch_vit_history[
        "val_loss"
    ].append(
        val_loss
    )

    pytorch_vit_history[
        "train_accuracy"
    ].append(
        train_accuracy
    )

    pytorch_vit_history[
        "val_accuracy"
    ].append(
        val_accuracy
    )


    # ========================================================
    # UPDATE LEARNING RATE
    # ========================================================

    vit_scheduler.step(
        val_loss
    )


    # ========================================================
    # CURRENT LEARNING RATE
    # ========================================================

    current_lr = (
        vit_optimizer.param_groups[0]["lr"]
    )


    # ========================================================
    # SAVE BEST MODEL
    # ========================================================

    if val_loss < best_val_loss:

        best_val_loss = val_loss

        patience_counter = 0

        torch.save(
            pytorch_vit.state_dict(),
            BEST_VIT_PATH
        )

        status = "  BEST"

    else:

        patience_counter += 1

        status = ""


    # ========================================================
    # EPOCH TIME
    # ========================================================

    epoch_time = (
        time.time() -
        epoch_start
    )


    # ========================================================
    # PRINT RESULTS
    # ========================================================

    print(
        f"Epoch [{epoch + 1:02d}/{EPOCHS}] | "
        f"Train Loss: {train_loss:.4f} | "
        f"Train Acc: {train_accuracy:.4f} | "
        f"Val Loss: {val_loss:.4f} | "
        f"Val Acc: {val_accuracy:.4f} | "
        f"LR: {current_lr:.2e} | "
        f"Time: {epoch_time:.1f}s"
        f"{status}"
    )


    # ========================================================
    # EARLY STOPPING
    # ========================================================

    if patience_counter >= PATIENCE:

        print(
            "\n Early stopping triggered."
        )

        break


# ============================================================
# LOAD BEST MODEL
# ============================================================

if os.path.exists(BEST_VIT_PATH):

    pytorch_vit.load_state_dict(
        torch.load(
            BEST_VIT_PATH,
            map_location=device
        )
    )

    print(
        "\n Best PyTorch ViT weights restored."
    )


# ============================================================
# COMPLETE
# ============================================================

print("\n" + "=" * 80)
print(" PYTORCH ViT TRAINING COMPLETE")
print("=" * 80)

print("\nBest model:")
print(BEST_VIT_PATH)

print(
    f"\nBest validation loss: "
    f"{best_val_loss:.4f}"
)

print(
    f"\nEpochs completed: "
    f"{len(pytorch_vit_history['train_loss'])}"
)
````

````python
# ============================================================
# STEP 38 — CNN + VISION TRANSFORMER HYBRID
# ============================================================

import os
import torch
import torch.nn as nn


# ============================================================
# CONFIGURATION
# ============================================================

IMAGE_SIZE = 64
PATCH_SIZE = 8

IN_CHANNELS = 3

NUM_PATCHES = (
    IMAGE_SIZE // PATCH_SIZE
) ** 2

CNN_FEATURES = 128

VIT_EMBED_DIM = 64

VIT_HEADS = 4

VIT_LAYERS = 3

VIT_MLP_DIM = 128

FUSION_DIM = 128


# ============================================================
# HYBRID MODEL
# ============================================================

class CNNViTHybrid(nn.Module):

    def __init__(self):

        super().__init__()


        # ====================================================
        # CNN BRANCH
        # ====================================================

        self.cnn_branch = nn.Sequential(

            # -----------------------------------------------
            # Block 1
            # -----------------------------------------------

            nn.Conv2d(
                3,
                32,
                kernel_size=3,
                padding=1
            ),

            nn.BatchNorm2d(32),

            nn.ReLU(),

            nn.MaxPool2d(2),

            # 64 → 32


            # -----------------------------------------------
            # Block 2
            # -----------------------------------------------

            nn.Conv2d(
                32,
                64,
                kernel_size=3,
                padding=1
            ),

            nn.BatchNorm2d(64),

            nn.ReLU(),

            nn.MaxPool2d(2),

            # 32 → 16


            # -----------------------------------------------
            # Block 3
            # -----------------------------------------------

            nn.Conv2d(
                64,
                128,
                kernel_size=3,
                padding=1
            ),

            nn.BatchNorm2d(128),

            nn.ReLU(),

            nn.MaxPool2d(2),

            # 16 → 8


            # -----------------------------------------------
            # Global pooling
            # -----------------------------------------------

            nn.AdaptiveAvgPool2d(
                (1, 1)
            )
        )


        # ====================================================
        # ViT PATCH EMBEDDING
        # ====================================================

        self.patch_embedding = nn.Conv2d(
            in_channels=IN_CHANNELS,
            out_channels=VIT_EMBED_DIM,
            kernel_size=PATCH_SIZE,
            stride=PATCH_SIZE
        )


        # ====================================================
        # CLASS TOKEN
        # ====================================================

        self.cls_token = nn.Parameter(
            torch.zeros(
                1,
                1,
                VIT_EMBED_DIM
            )
        )


        # ====================================================
        # POSITION EMBEDDING
        # ====================================================

        self.position_embedding = nn.Parameter(
            torch.zeros(
                1,
                NUM_PATCHES + 1,
                VIT_EMBED_DIM
            )
        )


        # ====================================================
        # TRANSFORMER
        # ====================================================

        encoder_layer = (
            nn.TransformerEncoderLayer(

                d_model=VIT_EMBED_DIM,

                nhead=VIT_HEADS,

                dim_feedforward=VIT_MLP_DIM,

                dropout=0.10,

                activation="gelu",

                batch_first=True,

                norm_first=True
            )
        )


        self.transformer = (
            nn.TransformerEncoder(
                encoder_layer,
                num_layers=VIT_LAYERS
            )
        )


        # ====================================================
        # ViT NORMALIZATION
        # ====================================================

        self.vit_norm = nn.LayerNorm(
            VIT_EMBED_DIM
        )


        # ====================================================
        # FUSION LAYER
        # ====================================================

        # CNN feature = 128
        # ViT feature = 64
        #
        # Combined = 192

        self.fusion = nn.Sequential(

            nn.Linear(
                CNN_FEATURES + VIT_EMBED_DIM,
                FUSION_DIM
            ),

            nn.BatchNorm1d(
                FUSION_DIM
            ),

            nn.ReLU(),

            nn.Dropout(
                0.30
            )
        )


        # ====================================================
        # FINAL CLASSIFIER
        # ====================================================

        self.classifier = nn.Sequential(

            nn.Linear(
                FUSION_DIM,
                64
            ),

            nn.ReLU(),

            nn.Dropout(
                0.25
            ),

            nn.Linear(
                64,
                1
            )
        )


    # ========================================================
    # FORWARD
    # ========================================================

    def forward(self, x):


        # ====================================================
        # CNN BRANCH
        # ====================================================

        cnn_features = self.cnn_branch(
            x
        )

        # Shape:
        #
        # [batch, 128, 1, 1]

        cnn_features = cnn_features.flatten(
            1
        )

        # Shape:
        #
        # [batch, 128]


        # ====================================================
        # ViT BRANCH
        # ====================================================

        patches = self.patch_embedding(
            x
        )

        # Shape:
        #
        # [batch, 64, 8, 8]


        # Flatten patches

        patches = patches.flatten(
            2
        )

        # [batch, 64, 64]


        # Move embedding dimension

        patches = patches.transpose(
            1,
            2
        )

        # [batch, 64, 64]


        # ====================================================
        # ADD CLS TOKEN
        # ====================================================

        batch_size = patches.size(0)

        cls_tokens = self.cls_token.expand(
            batch_size,
            -1,
            -1
        )

        tokens = torch.cat(
            [
                cls_tokens,
                patches
            ],
            dim=1
        )

        # [batch, 65, 64]


        # ====================================================
        # POSITION EMBEDDING
        # ====================================================

        tokens = (
            tokens
            +
            self.position_embedding
        )


        # ====================================================
        # TRANSFORMER
        # ====================================================

        transformer_output = self.transformer(
            tokens
        )


        # ====================================================
        # CLS FEATURE
        # ====================================================

        vit_features = transformer_output[
            :,
            0
        ]

        # [batch, 64]


        # Normalize

        vit_features = self.vit_norm(
            vit_features
        )


        # ====================================================
        # FUSION
        # ====================================================

        combined_features = torch.cat(
            [
                cnn_features,
                vit_features
            ],
            dim=1
        )

        # [batch, 192]


        fused_features = self.fusion(
            combined_features
        )

        # [batch, 128]


        # ====================================================
        # CLASSIFICATION
        # ====================================================

        output = self.classifier(
            fused_features
        )

        return output


# ============================================================
# CREATE MODEL
# ============================================================

hybrid_model = CNNViTHybrid()

hybrid_model = hybrid_model.to(
    device
)


# ============================================================
# LOSS
# ============================================================

hybrid_criterion = nn.BCEWithLogitsLoss()


# ============================================================
# OPTIMIZER
# ============================================================

hybrid_optimizer = torch.optim.AdamW(
    hybrid_model.parameters(),
    lr=3e-4,
    weight_decay=1e-4
)


# ============================================================
# MODEL INFORMATION
# ============================================================

total_parameters = sum(
    p.numel()
    for p in hybrid_model.parameters()
)

trainable_parameters = sum(
    p.numel()
    for p in hybrid_model.parameters()
    if p.requires_grad
)


print("=" * 80)
print("CNN + ViT HYBRID MODEL")
print("=" * 80)

print(
    f"\nImage size       : {IMAGE_SIZE} × {IMAGE_SIZE}"
)

print(
    f"Patch size       : {PATCH_SIZE} × {PATCH_SIZE}"
)

print(
    f"Number patches   : {NUM_PATCHES}"
)

print(
    f"CNN features     : {CNN_FEATURES}"
)

print(
    f"ViT embedding    : {VIT_EMBED_DIM}"
)

print(
    f"Fusion dimension : {FUSION_DIM}"
)

print(
    f"\nTotal parameters : {total_parameters:,}"
)

print(
    f"Trainable params : {trainable_parameters:,}"
)

print(
    f"Device           : {device}"
)

print("\n" + "=" * 80)
print(" CNN + ViT HYBRID CREATED")
print("=" * 80)
````

````python
# ============================================================
# STEP 38 — CNN + VISION TRANSFORMER HYBRID
# ============================================================

import os
import torch
import torch.nn as nn


# ============================================================
# CONFIGURATION
# ============================================================

IMAGE_SIZE = 64
PATCH_SIZE = 8

IN_CHANNELS = 3

NUM_PATCHES = (
    IMAGE_SIZE // PATCH_SIZE
) ** 2

CNN_FEATURES = 128

VIT_EMBED_DIM = 64

VIT_HEADS = 4

VIT_LAYERS = 3

VIT_MLP_DIM = 128

FUSION_DIM = 128


# ============================================================
# HYBRID MODEL
# ============================================================

class CNNViTHybrid(nn.Module):

    def __init__(self):

        super().__init__()


        # ====================================================
        # CNN BRANCH
        # ====================================================

        self.cnn_branch = nn.Sequential(

            # -----------------------------------------------
            # Block 1
            # -----------------------------------------------

            nn.Conv2d(
                3,
                32,
                kernel_size=3,
                padding=1
            ),

            nn.BatchNorm2d(32),

            nn.ReLU(),

            nn.MaxPool2d(2),

            # 64 → 32


            # -----------------------------------------------
            # Block 2
            # -----------------------------------------------

            nn.Conv2d(
                32,
                64,
                kernel_size=3,
                padding=1
            ),

            nn.BatchNorm2d(64),

            nn.ReLU(),

            nn.MaxPool2d(2),

            # 32 → 16


            # -----------------------------------------------
            # Block 3
            # -----------------------------------------------

            nn.Conv2d(
                64,
                128,
                kernel_size=3,
                padding=1
            ),

            nn.BatchNorm2d(128),

            nn.ReLU(),

            nn.MaxPool2d(2),

            # 16 → 8


            # -----------------------------------------------
            # Global pooling
            # -----------------------------------------------

            nn.AdaptiveAvgPool2d(
                (1, 1)
            )
        )


        # ====================================================
        # ViT PATCH EMBEDDING
        # ====================================================

        self.patch_embedding = nn.Conv2d(
            in_channels=IN_CHANNELS,
            out_channels=VIT_EMBED_DIM,
            kernel_size=PATCH_SIZE,
            stride=PATCH_SIZE
        )


        # ====================================================
        # CLASS TOKEN
        # ====================================================

        self.cls_token = nn.Parameter(
            torch.zeros(
                1,
                1,
                VIT_EMBED_DIM
            )
        )


        # ====================================================
        # POSITION EMBEDDING
        # ====================================================

        self.position_embedding = nn.Parameter(
            torch.zeros(
                1,
                NUM_PATCHES + 1,
                VIT_EMBED_DIM
            )
        )


        # ====================================================
        # TRANSFORMER
        # ====================================================

        encoder_layer = (
            nn.TransformerEncoderLayer(

                d_model=VIT_EMBED_DIM,

                nhead=VIT_HEADS,

                dim_feedforward=VIT_MLP_DIM,

                dropout=0.10,

                activation="gelu",

                batch_first=True,

                norm_first=True
            )
        )


        self.transformer = (
            nn.TransformerEncoder(
                encoder_layer,
                num_layers=VIT_LAYERS
            )
        )


        # ====================================================
        # ViT NORMALIZATION
        # ====================================================

        self.vit_norm = nn.LayerNorm(
            VIT_EMBED_DIM
        )


        # ====================================================
        # FUSION LAYER
        # ====================================================

        # CNN feature = 128
        # ViT feature = 64
        #
        # Combined = 192

        self.fusion = nn.Sequential(

            nn.Linear(
                CNN_FEATURES + VIT_EMBED_DIM,
                FUSION_DIM
            ),

            nn.BatchNorm1d(
                FUSION_DIM
            ),

            nn.ReLU(),

            nn.Dropout(
                0.30
            )
        )


        # ====================================================
        # FINAL CLASSIFIER
        # ====================================================

        self.classifier = nn.Sequential(

            nn.Linear(
                FUSION_DIM,
                64
            ),

            nn.ReLU(),

            nn.Dropout(
                0.25
            ),

            nn.Linear(
                64,
                1
            )
        )


    # ========================================================
    # FORWARD
    # ========================================================

    def forward(self, x):


        # ====================================================
        # CNN BRANCH
        # ====================================================

        cnn_features = self.cnn_branch(
            x
        )

        # Shape:
        #
        # [batch, 128, 1, 1]

        cnn_features = cnn_features.flatten(
            1
        )

        # Shape:
        #
        # [batch, 128]


        # ====================================================
        # ViT BRANCH
        # ====================================================

        patches = self.patch_embedding(
            x
        )

        # Shape:
        #
        # [batch, 64, 8, 8]


        # Flatten patches

        patches = patches.flatten(
            2
        )

        # [batch, 64, 64]


        # Move embedding dimension

        patches = patches.transpose(
            1,
            2
        )

        # [batch, 64, 64]


        # ====================================================
        # ADD CLS TOKEN
        # ====================================================

        batch_size = patches.size(0)

        cls_tokens = self.cls_token.expand(
            batch_size,
            -1,
            -1
        )

        tokens = torch.cat(
            [
                cls_tokens,
                patches
            ],
            dim=1
        )

        # [batch, 65, 64]


        # ====================================================
        # POSITION EMBEDDING
        # ====================================================

        tokens = (
            tokens
            +
            self.position_embedding
        )


        # ====================================================
        # TRANSFORMER
        # ====================================================

        transformer_output = self.transformer(
            tokens
        )


        # ====================================================
        # CLS FEATURE
        # ====================================================

        vit_features = transformer_output[
            :,
            0
        ]

        # [batch, 64]


        # Normalize

        vit_features = self.vit_norm(
            vit_features
        )


        # ====================================================
        # FUSION
        # ====================================================

        combined_features = torch.cat(
            [
                cnn_features,
                vit_features
            ],
            dim=1
        )

        # [batch, 192]


        fused_features = self.fusion(
            combined_features
        )

        # [batch, 128]


        # ====================================================
        # CLASSIFICATION
        # ====================================================

        output = self.classifier(
            fused_features
        )

        return output


# ============================================================
# CREATE MODEL
# ============================================================

hybrid_model = CNNViTHybrid()

hybrid_model = hybrid_model.to(
    device
)


# ============================================================
# LOSS
# ============================================================

hybrid_criterion = nn.BCEWithLogitsLoss()


# ============================================================
# OPTIMIZER
# ============================================================

hybrid_optimizer = torch.optim.AdamW(
    hybrid_model.parameters(),
    lr=3e-4,
    weight_decay=1e-4
)


# ============================================================
# MODEL INFORMATION
# ============================================================

total_parameters = sum(
    p.numel()
    for p in hybrid_model.parameters()
)

trainable_parameters = sum(
    p.numel()
    for p in hybrid_model.parameters()
    if p.requires_grad
)


print("=" * 80)
print("CNN + ViT HYBRID MODEL")
print("=" * 80)

print(
    f"\nImage size       : {IMAGE_SIZE} × {IMAGE_SIZE}"
)

print(
    f"Patch size       : {PATCH_SIZE} × {PATCH_SIZE}"
)

print(
    f"Number patches   : {NUM_PATCHES}"
)

print(
    f"CNN features     : {CNN_FEATURES}"
)

print(
    f"ViT embedding    : {VIT_EMBED_DIM}"
)

print(
    f"Fusion dimension : {FUSION_DIM}"
)

print(
    f"\nTotal parameters : {total_parameters:,}"
)

print(
    f"Trainable params : {trainable_parameters:,}"
)

print(
    f"Device           : {device}"
)

print("\n" + "=" * 80)
print(" CNN + ViT HYBRID CREATED")
print("=" * 80)
````

````python
# ============================================================
# STEP 39 — TRAIN CNN + ViT HYBRID
# ============================================================

import os
import time
import torch


# ============================================================
# CONFIGURATION
# ============================================================

EPOCHS = 20

PATIENCE = 5

BASE_DIR = (
    "/content/Agricultural_Land_Classification"
)

MODEL_DIR = os.path.join(
    BASE_DIR,
    "models"
)

os.makedirs(
    MODEL_DIR,
    exist_ok=True
)

BEST_HYBRID_PATH = os.path.join(
    MODEL_DIR,
    "cnn_vit_hybrid_best.pth"
)


# ============================================================
# HISTORY
# ============================================================

hybrid_history = {

    "train_loss": [],

    "val_loss": [],

    "train_accuracy": [],

    "val_accuracy": []
}


# ============================================================
# BEST MODEL
# ============================================================

best_val_loss = float(
    "inf"
)

patience_counter = 0


# ============================================================
# LEARNING RATE SCHEDULER
# ============================================================

hybrid_scheduler = (
    torch.optim.lr_scheduler.ReduceLROnPlateau(

        hybrid_optimizer,

        mode="min",

        factor=0.5,

        patience=2
    )
)


# ============================================================
# TRAINING
# ============================================================

print("=" * 80)
print(" TRAINING CNN + ViT HYBRID")
print("=" * 80)

print(
    f"\nDevice : {device}"
)

print(
    f"Epochs : {EPOCHS}"
)

print(
    f"Early stopping patience : {PATIENCE}"
)

print("\n" + "=" * 80)


for epoch in range(EPOCHS):

    start_time = time.time()


    # ========================================================
    # TRAIN
    # ========================================================

    hybrid_model.train()

    train_loss_total = 0.0

    train_correct = 0

    train_total = 0


    for images, labels in train_loader:

        images = images.to(
            device,
            non_blocking=True
        )

        labels = labels.to(
            device,
            non_blocking=True
        )

        labels = labels.float().unsqueeze(
            1
        )


        # ----------------------------------------------------
        # Clear gradients
        # ----------------------------------------------------

        hybrid_optimizer.zero_grad()


        # ----------------------------------------------------
        # Forward
        # ----------------------------------------------------

        outputs = hybrid_model(
            images
        )


        # ----------------------------------------------------
        # Loss
        # ----------------------------------------------------

        loss = hybrid_criterion(
            outputs,
            labels
        )


        # ----------------------------------------------------
        # Backprop
        # ----------------------------------------------------

        loss.backward()


        # ----------------------------------------------------
        # Update
        # ----------------------------------------------------

        hybrid_optimizer.step()


        # ----------------------------------------------------
        # Statistics
        # ----------------------------------------------------

        batch_size = labels.size(0)

        train_loss_total += (
            loss.item()
            *
            batch_size
        )


        probabilities = torch.sigmoid(
            outputs
        )

        predictions = (
            probabilities >= 0.5
        ).float()


        train_correct += (
            predictions == labels
        ).sum().item()


        train_total += batch_size


    # ========================================================
    # TRAIN METRICS
    # ========================================================

    train_loss = (
        train_loss_total
        /
        train_total
    )

    train_accuracy = (
        train_correct
        /
        train_total
    )


    # ========================================================
    # VALIDATION
    # ========================================================

    hybrid_model.eval()

    val_loss_total = 0.0

    val_correct = 0

    val_total = 0


    with torch.no_grad():

        for images, labels in val_loader:

            images = images.to(
                device,
                non_blocking=True
            )

            labels = labels.to(
                device,
                non_blocking=True
            )

            labels = labels.float().unsqueeze(
                1
            )


            outputs = hybrid_model(
                images
            )


            loss = hybrid_criterion(
                outputs,
                labels
            )


            batch_size = labels.size(0)

            val_loss_total += (
                loss.item()
                *
                batch_size
            )


            probabilities = torch.sigmoid(
                outputs
            )

            predictions = (
                probabilities >= 0.5
            ).float()


            val_correct += (
                predictions == labels
            ).sum().item()


            val_total += batch_size


    # ========================================================
    # VALIDATION METRICS
    # ========================================================

    val_loss = (
        val_loss_total
        /
        val_total
    )

    val_accuracy = (
        val_correct
        /
        val_total
    )


    # ========================================================
    # HISTORY
    # ========================================================

    hybrid_history[
        "train_loss"
    ].append(
        train_loss
    )

    hybrid_history[
        "val_loss"
    ].append(
        val_loss
    )

    hybrid_history[
        "train_accuracy"
    ].append(
        train_accuracy
    )

    hybrid_history[
        "val_accuracy"
    ].append(
        val_accuracy
    )


    # ========================================================
    # SCHEDULER
    # ========================================================

    hybrid_scheduler.step(
        val_loss
    )


    current_lr = (
        hybrid_optimizer
        .param_groups[0]["lr"]
    )


    # ========================================================
    # SAVE BEST MODEL
    # ========================================================

    if val_loss < best_val_loss:

        best_val_loss = val_loss

        patience_counter = 0

        torch.save(
            hybrid_model.state_dict(),
            BEST_HYBRID_PATH
        )

        status = "  BEST"

    else:

        patience_counter += 1

        status = ""


    # ========================================================
    # TIME
    # ========================================================

    elapsed = (
        time.time()
        -
        start_time
    )


    # ========================================================
    # PRINT
    # ========================================================

    print(
        f"Epoch [{epoch + 1:02d}/{EPOCHS}] | "
        f"Train Loss: {train_loss:.4f} | "
        f"Train Acc: {train_accuracy:.4f} | "
        f"Val Loss: {val_loss:.4f} | "
        f"Val Acc: {val_accuracy:.4f} | "
        f"LR: {current_lr:.2e} | "
        f"Time: {elapsed:.1f}s"
        f"{status}"
    )


    # ========================================================
    # EARLY STOPPING
    # ========================================================

    if patience_counter >= PATIENCE:

        print(
            "\n Early stopping triggered."
        )

        break


# ============================================================
# RESTORE BEST MODEL
# ============================================================

if os.path.exists(
    BEST_HYBRID_PATH
):

    hybrid_model.load_state_dict(
        torch.load(
            BEST_HYBRID_PATH,
            map_location=device
        )
    )

    print(
        "\n Best hybrid model restored."
    )


# ============================================================
# COMPLETE
# ============================================================

print("\n" + "=" * 80)
print("CNN + ViT HYBRID TRAINING COMPLETE")
print("=" * 80)

print(
    f"\nBest validation loss: "
    f"{best_val_loss:.4f}"
)

print(
    f"Epochs completed: "
    f"{len(hybrid_history['train_loss'])}"
)

print(
    f"\nSaved model:"
)

print(
    BEST_HYBRID_PATH
)
````

````python
# ============================================================
# STEP 46 — CONFUSION MATRIX + CLASSIFICATION REPORT
# FIXED FOR CUDA / PYTORCH TENSORS
# ============================================================

import os
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

from sklearn.metrics import (
    confusion_matrix,
    classification_report,
    accuracy_score,
    precision_score,
    recall_score,
    f1_score
)

# ============================================================
# PATHS
# ============================================================

BASE_DIR = "/content/Agricultural_Land_Classification"

METRICS_DIR = os.path.join(
    BASE_DIR,
    "results",
    "metrics"
)

FIGURES_DIR = os.path.join(
    BASE_DIR,
    "results",
    "figures"
)

os.makedirs(METRICS_DIR, exist_ok=True)
os.makedirs(FIGURES_DIR, exist_ok=True)

# ============================================================
# LOAD LEADERBOARD
# ============================================================

leaderboard_path = os.path.join(
    METRICS_DIR,
    "FINAL_MODEL_LEADERBOARD.csv"
)

if not os.path.exists(leaderboard_path):
    raise FileNotFoundError(
        "FINAL_MODEL_LEADERBOARD.csv not found. "
        "Run STEP 44 first."
    )

leaderboard = pd.read_csv(
    leaderboard_path
)

# ============================================================
# BEST MODEL
# ============================================================

if "F1" not in leaderboard.columns:
    raise ValueError(
        "F1 column not found in leaderboard."
    )

best_row = leaderboard.loc[
    leaderboard["F1"].idxmax()
]

BEST_MODEL_NAME = best_row["Model"]

print("=" * 90)
print("🔍 STEP 46 — BEST MODEL ERROR ANALYSIS")
print("=" * 90)

print(
    f"\n🏆 Best Model: {BEST_MODEL_NAME}"
)

print(
    f"F1 Score: {best_row['F1']:.4f}"
)

# ============================================================
# SAFE TENSOR → NUMPY CONVERTER
# ============================================================

def to_numpy_safe(value):

    # PyTorch tensor
    if hasattr(value, "detach"):

        return (
            value
            .detach()
            .cpu()
            .numpy()
            .flatten()
        )

    # NumPy / list
    return np.asarray(
        value
    ).flatten()


# ============================================================
# FIND TRUE LABELS
# ============================================================

candidate_true_names = [
    "true_labels",
    "y_true",
    "test_labels",
    "y_test",
    "y_test_true",
    "actual_labels"
]

candidate_pred_names = [
    "pred_labels",
    "predicted_labels",
    "y_pred",
    "y_test_pred",
    "test_predictions",
    "predictions",
    "test_preds",
    "y_pred_labels"
]

y_true = None
y_pred = None

true_source = None
pred_source = None

# ============================================================
# SEARCH TRUE LABELS
# ============================================================

for name in candidate_true_names:

    if name in globals():

        value = globals()[name]

        if value is not None:

            try:

                y_true = to_numpy_safe(
                    value
                )

                true_source = name

                print(
                    f"\n✅ Found true labels: {name}"
                )

                print(
                    f"   Shape: {y_true.shape}"
                )

                break

            except Exception as e:

                print(
                    f"Could not use {name}: {e}"
                )


# ============================================================
# SEARCH PREDICTIONS
# ============================================================

for name in candidate_pred_names:

    if name in globals():

        value = globals()[name]

        if value is not None:

            try:

                y_pred = to_numpy_safe(
                    value
                )

                pred_source = name

                print(
                    f"✅ Found predictions: {name}"
                )

                print(
                    f"   Shape: {y_pred.shape}"
                )

                break

            except Exception as e:

                print(
                    f"Could not use {name}: {e}"
                )


# ============================================================
# CHECK TRUE LABELS
# ============================================================

if y_true is None:

    raise ValueError(
        """
True labels were not found.

Please run the previous test/evaluation cell
that creates the test labels.
"""
    )


# ============================================================
# CHECK PREDICTIONS
# ============================================================

if y_pred is None:

    print("\n⚠️ Predictions were not found.")

    print(
        """
Your true labels were found successfully,
but this notebook does not currently have
the predicted labels available.

We will need the prediction variable from
the model evaluation step before creating
the confusion matrix.
"""
    )

else:

    # ========================================================
    # PROCESS PREDICTIONS
    # ========================================================

    y_pred = np.asarray(
        y_pred
    )

    # --------------------------------------------------------
    # Probability / logits handling
    # --------------------------------------------------------

    if y_pred.ndim > 1:

        if y_pred.shape[-1] == 1:

            y_pred = (
                y_pred.ravel() >= 0.5
            ).astype(int)

        else:

            y_pred = np.argmax(
                y_pred,
                axis=1
            )

    else:

        # If values appear to be probabilities
        if (
            np.issubdtype(
                y_pred.dtype,
                np.floating
            )
            and
            np.all(
                (y_pred >= 0)
                &
                (y_pred <= 1)
            )
            and
            np.any(
                (y_pred > 0)
                &
                (y_pred < 1)
            )
        ):

            y_pred = (
                y_pred >= 0.5
            ).astype(int)

        else:

            y_pred = y_pred.astype(int)

    # ========================================================
    # LENGTH CHECK
    # ========================================================

    if len(y_true) != len(y_pred):

        raise ValueError(
            f"""
Length mismatch!

True labels : {len(y_true)}
Predictions : {len(y_pred)}

The labels and predictions must belong
to the same test set.
"""
        )

    # ========================================================
    # CLASS NAMES
    # ========================================================

    class_names = [
        "Non-Agricultural Land",
        "Agricultural Land"
    ]

    # ========================================================
    # CONFUSION MATRIX
    # ========================================================

    cm = confusion_matrix(
        y_true,
        y_pred,
        labels=[0, 1]
    )

    TN = cm[0, 0]
    FP = cm[0, 1]
    FN = cm[1, 0]
    TP = cm[1, 1]

    print("\n" + "=" * 90)
    print("📊 CONFUSION MATRIX")
    print("=" * 90)

    print(
        "\nRows    = Actual"
    )

    print(
        "Columns = Predicted"
    )

    print("\n", cm)

    print("\nDetailed Results:")

    print(
        f"True Negatives  (TN): {TN}"
    )

    print(
        f"False Positives (FP): {FP}"
    )

    print(
        f"False Negatives (FN): {FN}"
    )

    print(
        f"True Positives  (TP): {TP}"
    )

    # ========================================================
    # CONFUSION MATRIX VISUALIZATION
    # ========================================================

    plt.figure(
        figsize=(9, 8)
    )

    plt.imshow(
        cm,
        interpolation="nearest"
    )

    plt.title(
        f"Confusion Matrix — {BEST_MODEL_NAME}",
        fontsize=16,
        fontweight="bold"
    )

    plt.colorbar(
        label="Number of Images"
    )

    tick_marks = np.arange(2)

    plt.xticks(
        tick_marks,
        class_names,
        rotation=20,
        ha="right"
    )

    plt.yticks(
        tick_marks,
        class_names
    )

    plt.xlabel(
        "Predicted Label",
        fontsize=12
    )

    plt.ylabel(
        "Actual Label",
        fontsize=12
    )

    for i in range(2):

        for j in range(2):

            plt.text(
                j,
                i,
                str(cm[i, j]),
                ha="center",
                va="center",
                fontsize=18,
                fontweight="bold"
            )

    plt.tight_layout()

    confusion_figure = os.path.join(
        FIGURES_DIR,
        "BEST_MODEL_CONFUSION_MATRIX.png"
    )

    plt.savefig(
        confusion_figure,
        dpi=200,
        bbox_inches="tight"
    )

    plt.show()

    # ========================================================
    # CLASSIFICATION REPORT
    # ========================================================

    report = classification_report(
        y_true,
        y_pred,
        labels=[0, 1],
        target_names=class_names,
        output_dict=True,
        zero_division=0
    )

    report_df = pd.DataFrame(
        report
    ).transpose()

    print("\n" + "=" * 90)
    print(" CLASSIFICATION REPORT")
    print("=" * 90)

    display(
        report_df.style.format(
            {
                "precision": "{:.4f}",
                "recall": "{:.4f}",
                "f1-score": "{:.4f}",
                "support": "{:.0f}"
            },
            na_rep="N/A"
        )
    )

    # ========================================================
    # OVERALL METRICS
    # ========================================================

    accuracy = accuracy_score(
        y_true,
        y_pred
    )

    precision = precision_score(
        y_true,
        y_pred,
        zero_division=0
    )

    recall = recall_score(
        y_true,
        y_pred,
        zero_division=0
    )

    f1 = f1_score(
        y_true,
        y_pred,
        zero_division=0
    )

    print("\n" + "=" * 90)
    print(" FINAL TEST PERFORMANCE")
    print("=" * 90)

    print(
        f"Accuracy  : {accuracy:.4f}"
    )

    print(
        f"Precision : {precision:.4f}"
    )

    print(
        f"Recall    : {recall:.4f}"
    )

    print(
        f"F1 Score  : {f1:.4f}"
    )

    # ========================================================
    # SAVE CLASSIFICATION REPORT
    # ========================================================

    report_path = os.path.join(
        METRICS_DIR,
        "BEST_MODEL_CLASSIFICATION_REPORT.csv"
    )

    report_df.to_csv(
        report_path
    )

    # ========================================================
    # SAVE CONFUSION MATRIX
    # ========================================================

    cm_df = pd.DataFrame(
        cm,
        index=[
            "Actual_Non_Agricultural",
            "Actual_Agricultural"
        ],
        columns=[
            "Predicted_Non_Agricultural",
            "Predicted_Agricultural"
        ]
    )

    cm_path = os.path.join(
        METRICS_DIR,
        "BEST_MODEL_CONFUSION_MATRIX.csv"
    )

    cm_df.to_csv(
        cm_path
    )

    # ========================================================
    # ERROR ANALYSIS
    # ========================================================

    print("\n" + "=" * 90)
    print(" ERROR ANALYSIS")
    print("=" * 90)

    print(
        f"""
The best model is:

    {BEST_MODEL_NAME}

Correct classifications:
    • Non-agricultural → Non-agricultural: {TN}
    • Agricultural → Agricultural:         {TP}

Incorrect classifications:
    • Non-agricultural → Agricultural:     {FP}
    • Agricultural → Non-agricultural:     {FN}

The confusion matrix shows exactly where
the model succeeds and where it makes mistakes.
"""
    )

    # ========================================================
    # SAVE PREDICTIONS
    # ========================================================

    predictions_df = pd.DataFrame(
        {
            "Actual": y_true,
            "Predicted": y_pred
        }
    )

    predictions_path = os.path.join(
        METRICS_DIR,
        "BEST_MODEL_PREDICTIONS.csv"
    )

    predictions_df.to_csv(
        predictions_path,
        index=False
    )

    # ========================================================
    # FINAL OUTPUT
    # ========================================================

    print("=" * 90)
    print(" SAVED OUTPUTS")
    print("=" * 90)

    print(
        f"1. {confusion_figure}"
    )

    print(
        f"2. {cm_path}"
    )

    print(
        f"3. {report_path}"
    )

    print(
        f"4. {predictions_path}"
    )

    print("\n" + "=" * 90)
    print("STEP 46 COMPLETE")
    print("=" * 90)
````

````python
# ============================================================
# STEP 47 — FIXED ROC-AUC ANALYSIS
# MATCH PROBABILITIES TO THE 900 TEST LABELS
# ============================================================

import os
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

from sklearn.metrics import (
    roc_curve,
    roc_auc_score
)

# ============================================================
# PATHS
# ============================================================

BASE_DIR = "/content/Agricultural_Land_Classification"

METRICS_DIR = os.path.join(
    BASE_DIR,
    "results",
    "metrics"
)

FIGURES_DIR = os.path.join(
    BASE_DIR,
    "results",
    "figures"
)

os.makedirs(
    METRICS_DIR,
    exist_ok=True
)

os.makedirs(
    FIGURES_DIR,
    exist_ok=True
)

# ============================================================
# LOAD TRUE LABELS FROM STEP 46
# ============================================================

prediction_file = os.path.join(
    METRICS_DIR,
    "BEST_MODEL_PREDICTIONS.csv"
)

if not os.path.exists(prediction_file):

    raise FileNotFoundError(
        "BEST_MODEL_PREDICTIONS.csv not found. "
        "Run STEP 46 first."
    )

predictions_df = pd.read_csv(
    prediction_file
)

y_true = predictions_df[
    "Actual"
].values.astype(int)

y_pred = predictions_df[
    "Predicted"
].values.astype(int)

EXPECTED_LENGTH = len(y_true)

# ============================================================
# LOAD BEST MODEL
# ============================================================

leaderboard_path = os.path.join(
    METRICS_DIR,
    "FINAL_MODEL_LEADERBOARD.csv"
)

leaderboard = pd.read_csv(
    leaderboard_path
)

best_row = leaderboard.loc[
    leaderboard["F1"].idxmax()
]

BEST_MODEL_NAME = best_row["Model"]

print("=" * 90)
print("📈 STEP 47 — ROC-AUC ANALYSIS")
print("=" * 90)

print(
    f"\n🏆 Best Model: {BEST_MODEL_NAME}"
)

print(
    f"Test labels: {EXPECTED_LENGTH}"
)

print(
    f"Leaderboard ROC-AUC: {best_row['ROC_AUC']:.4f}"
)

# ============================================================
# FUNCTION: CONVERT TO NUMPY
# ============================================================

def convert_to_numpy(value):

    try:

        if hasattr(value, "detach"):

            value = (
                value
                .detach()
                .cpu()
                .numpy()
            )

        else:

            value = np.asarray(
                value
            )

        return value

    except Exception:

        return None


# ============================================================
# SEARCH ALL EXISTING VARIABLES
# FOR A MATCHING PROBABILITY ARRAY
# ============================================================

print("\n🔎 Searching notebook variables...")
print(
    "Looking for probability arrays containing "
    f"{EXPECTED_LENGTH} test predictions."
)

probability_candidates = []

for name, value in list(
    globals().items()
):

    # Ignore private variables
    if name.startswith("_"):
        continue

    # Avoid searching obvious non-probability objects
    if name in [
        "y_true",
        "y_pred",
        "predictions_df",
        "leaderboard"
    ]:
        continue

    try:

        arr = convert_to_numpy(
            value
        )

        if arr is None:
            continue

        # Need at least one dimension
        if arr.ndim == 0:
            continue

        # ----------------------------------------------------
        # Possible shapes:
        #
        # (900,)
        # (900,1)
        # (900,2)
        # ----------------------------------------------------

        if arr.shape[0] == EXPECTED_LENGTH:

            if arr.ndim == 1:

                probability_candidates.append(
                    (
                        name,
                        arr.shape,
                        arr
                    )
                )

            elif arr.ndim == 2 and arr.shape[1] in [1, 2]:

                probability_candidates.append(
                    (
                        name,
                        arr.shape,
                        arr
                    )
                )

    except Exception:
        continue


# ============================================================
# DISPLAY CANDIDATES
# ============================================================

print(
    f"\nFound {len(probability_candidates)} "
    "matching candidate arrays."
)

for name, shape, arr in probability_candidates:

    print(
        f"  • {name}: {shape}"
    )

# ============================================================
# SELECT PROBABILITY ARRAY
# ============================================================

y_score = None
selected_name = None

# Prefer variables with probability-like names

preferred_keywords = [
    "prob",
    "score",
    "confidence",
    "pred"
]

for keyword in preferred_keywords:

    for name, shape, arr in probability_candidates:

        if keyword.lower() in name.lower():

            y_score = arr
            selected_name = name

            break

    if y_score is not None:
        break


# ============================================================
# IF NO MATCHING PROBABILITY ARRAY
# ============================================================

if y_score is None:

    print("\n" + "=" * 90)
    print("⚠️ MATCHING PROBABILITY SCORES NOT FOUND")
    print("=" * 90)

    print(
        """
The notebook contains the 900 test labels,
but no probability array with exactly 900
test predictions was found.

This means we should NOT create a fake ROC curve
from unrelated probabilities.

The verified leaderboard ROC-AUC is still:

    1.0000

We will use the actual CNN + ViT Hybrid test
probabilities if they are regenerated in the
next evaluation step.
"""
    )

else:

    # ========================================================
    # NORMALIZE SCORE ARRAY
    # ========================================================

    print(
        f"\n Using probability variable: "
        f"{selected_name}"
    )

    print(
        f"Original shape: {y_score.shape}"
    )

    y_score = np.asarray(
        y_score
    )

    # --------------------------------------------------------
    # Convert two-class probabilities
    # --------------------------------------------------------

    if y_score.ndim == 2:

        if y_score.shape[1] == 2:

            # Probability of class 1
            y_score = y_score[:, 1]

        elif y_score.shape[1] == 1:

            y_score = y_score[:, 0]

        else:

            raise ValueError(
                f"Unexpected probability shape: "
                f"{y_score.shape}"
            )

    # --------------------------------------------------------
    # Flatten
    # --------------------------------------------------------

    y_score = y_score.flatten()

    # ========================================================
    # FINAL LENGTH CHECK
    # ========================================================

    print(
        f"Final probability length: "
        f"{len(y_score)}"
    )

    print(
        f"True-label length: "
        f"{len(y_true)}"
    )

    if len(y_score) != len(y_true):

        raise ValueError(
            "Probability and label lengths still do not match."
        )

    # ========================================================
    # ROC-AUC
    # ========================================================

    roc_auc = roc_auc_score(
        y_true,
        y_score
    )

    fpr, tpr, thresholds = roc_curve(
        y_true,
        y_score
    )

    # ========================================================
    # PRINT RESULT
    # ========================================================

    print("\n" + "=" * 90)
    print(" VERIFIED ROC-AUC RESULT")
    print("=" * 90)

    print(
        f"Model   : {BEST_MODEL_NAME}"
    )

    print(
        f"ROC-AUC : {roc_auc:.4f}"
    )

    # ========================================================
    # ROC CURVE
    # ========================================================

    plt.figure(
        figsize=(9, 7)
    )

    plt.plot(
        fpr,
        tpr,
        linewidth=2,
        label=(
            f"{BEST_MODEL_NAME} "
            f"(AUC = {roc_auc:.4f})"
        )
    )

    plt.plot(
        [0, 1],
        [0, 1],
        linestyle="--",
        linewidth=1.5,
        label="Random Classifier"
    )

    plt.xlabel(
        "False Positive Rate",
        fontsize=12
    )

    plt.ylabel(
        "True Positive Rate",
        fontsize=12
    )

    plt.title(
        "ROC Curve — CNN + ViT Hybrid",
        fontsize=16,
        fontweight="bold"
    )

    plt.legend(
        loc="lower right"
    )

    plt.grid(
        alpha=0.3
    )

    plt.tight_layout()

    roc_figure_path = os.path.join(
        FIGURES_DIR,
        "BEST_MODEL_ROC_CURVE.png"
    )

    plt.savefig(
        roc_figure_path,
        dpi=200,
        bbox_inches="tight"
    )

    plt.show()

    # ========================================================
    # SAVE ROC DATA
    # ========================================================

    roc_data = pd.DataFrame(
        {
            "False_Positive_Rate": fpr,
            "True_Positive_Rate": tpr,
            "Threshold": thresholds
        }
    )

    roc_data_path = os.path.join(
        METRICS_DIR,
        "BEST_MODEL_ROC_DATA.csv"
    )

    roc_data.to_csv(
        roc_data_path,
        index=False
    )

    # ========================================================
    # SAVE ROC SUMMARY
    # ========================================================

    roc_summary = pd.DataFrame(
        {
            "Model": [
                BEST_MODEL_NAME
            ],
            "ROC_AUC": [
                roc_auc
            ],
            "Test_Samples": [
                EXPECTED_LENGTH
            ]
        }
    )

    roc_summary_path = os.path.join(
        METRICS_DIR,
        "BEST_MODEL_ROC_AUC.csv"
    )

    roc_summary.to_csv(
        roc_summary_path,
        index=False
    )

    # ========================================================
    # SAVE PROBABILITIES
    # ========================================================

    probability_df = pd.DataFrame(
        {
            "Actual": y_true,
            "Predicted": y_pred,
            "Probability_Class_1": y_score
        }
    )

    probability_path = os.path.join(
        METRICS_DIR,
        "BEST_MODEL_PROBABILITIES.csv"
    )

    probability_df.to_csv(
        probability_path,
        index=False
    )

    # ========================================================
    # OUTPUTS
    # ========================================================

    print("\n" + "=" * 90)
    print(" SAVED OUTPUTS")
    print("=" * 90)

    print(
        f"1. {roc_figure_path}"
    )

    print(
        f"2. {roc_data_path}"
    )

    print(
        f"3. {roc_summary_path}"
    )

    print(
        f"4. {probability_path}"
    )

print("\n" + "=" * 90)
print(" STEP 47 FINISHED")
print("=" * 90)
````

````python
# ============================================================
# STEP 48A — RECOVER CNN + ViT HYBRID MODEL
# ============================================================

import os
import torch
import torch.nn as nn

BASE_DIR = "/content/Agricultural_Land_Classification"

print("=" * 90)
print(" STEP 48A — SEARCHING FOR CNN + ViT HYBRID MODEL")
print("=" * 90)

# ============================================================
# 1. SEARCH CURRENT PYTHON MEMORY
# ============================================================

print("\n[1] Searching current notebook memory...")

found_model = None
found_name = None

for name, value in list(globals().items()):

    try:

        if isinstance(value, nn.Module):

            module_names = [
                n.lower()
                for n, _ in value.named_modules()
            ]

            model_name = name.lower()

            # Prefer hybrid / ViT models
            if (
                "hybrid" in model_name
                or
                "vit" in model_name
                or
                any(
                    "patch_embedding" in n
                    for n in module_names
                )
            ):

                found_model = value
                found_name = name

                break

    except Exception:
        pass


if found_model is not None:

    hybrid_model = found_model

    print(
        f" HYBRID MODEL RECOVERED FROM MEMORY"
    )

    print(
        f"Variable name: {found_name}"
    )

    print(
        f"Model type: {type(hybrid_model).__name__}"
    )

# ============================================================
# 2. SEARCH SAVED MODEL FILES
# ============================================================

if found_model is None:

    print(
        "\n[2] Model not found in memory."
    )

    print(
        "Searching project directory for saved PyTorch models..."
    )

    checkpoint_files = []

    for root, dirs, files in os.walk(
        BASE_DIR
    ):

        for filename in files:

            if filename.lower().endswith(
                (
                    ".pth",
                    ".pt",
                    ".bin"
                )
            ):

                checkpoint_files.append(
                    os.path.join(
                        root,
                        filename
                    )
                )

    if len(checkpoint_files) == 0:

        print(
            "\n❌ No .pth / .pt / .bin model files found."
        )

    else:

        print(
            f"\nFound {len(checkpoint_files)} model files:"
        )

        for i, filepath in enumerate(
            checkpoint_files,
            1
        ):

            size_mb = (
                os.path.getsize(filepath)
                / (1024 ** 2)
            )

            print(
                f"{i}. {filepath}"
            )

            print(
                f"   Size: {size_mb:.2f} MB"
            )

        # ====================================================
        # TRY TO LOAD CHECKPOINTS
        # ====================================================

        print(
            "\n[3] Inspecting checkpoint contents..."
        )

        for filepath in checkpoint_files:

            try:

                checkpoint = torch.load(
                    filepath,
                    map_location="cpu",
                    weights_only=False
                )

                print(
                    f"\n {os.path.basename(filepath)}"
                )

                print(
                    f"   Type: {type(checkpoint)}"
                )

                # ------------------------------------------------
                # Full saved model
                # ------------------------------------------------

                if isinstance(
                    checkpoint,
                    nn.Module
                ):

                    hybrid_model = checkpoint

                    found_model = checkpoint

                    print(
                        "    Full PyTorch model found!"
                    )

                    print(
                        f"   Model type: "
                        f"{type(checkpoint).__name__}"
                    )

                    break

                # ------------------------------------------------
                # Dictionary checkpoint
                # ------------------------------------------------

                elif isinstance(
                    checkpoint,
                    dict
                ):

                    print(
                        "   Dictionary checkpoint detected."
                    )

                    print(
                        "   Keys:"
                    )

                    for key in list(
                        checkpoint.keys()
                    )[:20]:

                        print(
                            f"      • {key}"
                        )

            except Exception as e:

                print(
                    f"   ⚠️ Could not load: {e}"
                )

# ============================================================
# 4. FINAL CHECK
# ============================================================

print("\n" + "=" * 90)

if found_model is not None:

    print(
        "✅ CNN + ViT HYBRID MODEL RECOVERED"
    )

    print("=" * 90)

    hybrid_model = found_model

    print(
        f"\nModel class:"
    )

    print(
        type(hybrid_model).__name__
    )

    print(
        "\nModel architecture:"
    )

    print(
        hybrid_model
    )

    # Move to GPU
    hybrid_model = hybrid_model.to(
        "cuda" if torch.cuda.is_available()
        else "cpu"
    )

    hybrid_model.eval()

    print(
        "\n✅ Model moved to device."
    )

    print(
        f"Device: "
        f"{next(hybrid_model.parameters()).device}"
    )

    print(
        "\n🎯 Ready for Grad-CAM."
    )

else:

    print(
        "❌ CNN + ViT HYBRID MODEL COULD NOT BE RECOVERED"
    )

    print("=" * 90)

    print(
        """
IMPORTANT:

Do NOT retrain the project yet.

The next step will identify the exact architecture/
checkpoint used earlier and restore the model correctly.
"""
    )
````
